# LinkedIn Tango Playground

Import the `Tango` object to solve the games bellow.

```python
from linkedin_games import Tango
```

## Tango No. [151](https://puzznest.com/linkedin-games/tango-151-answer)

March 7<sup>th</sup>, 2025

<div align='center'>
    <img src='./assets/tango-151.png' alt='Tango No. 151' width='30%'>
</div>

```python
tango = Tango(
    filled_squares={
        (1,2): 1,
        (1,5): 1,
        (5,2): 0,
        (5,5): 1
    },
    matching_pairs=[
        ((2,3), (2,4)),
        ((2,1), (3,1)),
        ((2,3), (3,3)),
        ((2,6), (3,6)),
        ((4,1), (4,2)),
        ((6,3), (6,4))
    ],
    opposite_pairs=[
        ((2,4), (3,4)),
        ((3,1), (4,1)),
        ((3,3), (3,4)),
        ((3,6), (4,6)),
        ((4,5), (4,6))
    ]
)
tango.solve()
tango.show()
```

Solution of Tango No. [151](https://puzznest.com/linkedin-games/tango-151-answer).
