```python
from linkedin_games import Zip
```

Zip No. 166

```python
numbered_squares = [
    (6,5), (6,6), (5,6), (6,4), (5,3), (4,2), (3,1), (2,1),
    (1,1), (1,2), (1,3), (3,5), (2,4), (3,4), (4,3), (4,6)
]
zip_game = Zip((6,6), numbered_squares)
zip_game.solve(verbose=True)
zip_game.show()
```

```python
numbered_squares = [(2,4), (5,2), (2,6), (1,7), (6,4), (3,6), (7,1), (6,2)]
walls = [
    ((1,2), (2,2)), ((1,3), (2,3)), ((2,1), (2,2)), ((2,4), (2,5)),
    ((3,1), (3,2)), ((3,4), (3,5)), ((3,3), (4,3)), ((3,4), (4,4)),
    ((4,4), (5,4)), ((4,5), (5,5)), ((5,3), (5,4)), ((5,6), (5,7)),
    ((6,3), (6,4)), ((6,6), (6,7)), ((6,5), (7,5)), ((6,6), (7,6))
]
zip_game = Zip((7,7), numbered_squares, walls)
zip_game.solve(verbose=True)
zip_game.show()
```
